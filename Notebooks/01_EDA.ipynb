STEP 3B — Fraud rate per categorical feature

```python
def fraud_rate_by_column(df, column):
    result = (
        df.groupby(column)["label"]
        .agg(
            total_claims="count",
            fraud_count="sum",
            fraud_rate="mean"
        )
        .sort_values("fraud_rate", ascending=False)
    )

    result["fraud_rate"] = (result["fraud_rate"] * 100).round(2)

    return result
```

```python
categorical_check = [
    "typeppk",
    "jkpst",
    "cmg",
    "diagprimer"
]

for col in categorical_check:
    print(f"\n{'='*60}")
    print(f"FRAUD RATE BY: {col}")
    print(f"{'='*60}")

    result = fraud_rate_by_column(train, col)
    display(result)
```

STEP 4A — Leakage check

```python
leakage_check = []

for col in train.columns:
    if col == "label":
        continue

    grouped = train.groupby(col)["label"].nunique()

    leakage_check.append({
        "column": col,
        "n_unique": train[col].nunique(),
        "n_values_with_both_labels": (grouped > 1).sum(),
        "n_values_with_single_label": (grouped == 1).sum()
    })

leakage_check = pd.DataFrame(leakage_check)

display(
    leakage_check.sort_values(
        "n_values_with_single_label",
        ascending=False
    )
)
```

STEP 4C — Cari nilai feature yang benar-benar 100% satu label

```python
feature_cols = [
    col for col in train.columns
    if col not in ["claim_id", "label"]
]

pure_value_results = []

for col in feature_cols:
    grouped = (
        train.groupby(col)["label"]
        .agg(
            count="count",
            fraud_rate="mean"
        )
        .reset_index()
    )

    grouped = grouped[grouped["count"] >= 100]

    pure = grouped[
        (grouped["fraud_rate"] == 0) |
        (grouped["fraud_rate"] == 1)
    ].copy()

    for _, row in pure.iterrows():
        pure_value_results.append({
            "column": col,
            "value": row[col],
            "count": row["count"],
            "fraud_rate": row["fraud_rate"]
        })

pure_value_results = pd.DataFrame(pure_value_results)

if len(pure_value_results) == 0:
    print("Tidak ditemukan nilai feature dengan >=100 observasi dan 100% satu label.")
else:
    pure_value_results = pure_value_results.sort_values(
        "count",
        ascending=False
    )

    display(pure_value_results)
```

STEP 5B — Fraud rate untuk seluruh kolom `dx2_*` dan `proc*`

```python
dx_proc_cols = [
    col for col in train.columns
    if col.startswith("dx2_") or col.startswith("proc")
]

dx_proc_summary = []

for col in dx_proc_cols:
    indicator = (train[col] > 0)

    rate_present = train.loc[indicator, "label"].mean() * 100 if indicator.sum() > 0 else np.nan
    rate_absent = train.loc[~indicator, "label"].mean() * 100 if (~indicator).sum() > 0 else np.nan
    n_present = indicator.sum()

    dx_proc_summary.append({
        "column": col,
        "n_present": n_present,
        "pct_present": round(n_present / len(train) * 100, 2),
        "fraud_rate_if_present": round(rate_present, 2) if pd.notna(rate_present) else np.nan,
        "fraud_rate_if_absent": round(rate_absent, 2) if pd.notna(rate_absent) else np.nan,
        "diff": round(rate_present - rate_absent, 2) if pd.notna(rate_present) and pd.notna(rate_absent) else np.nan
    })

dx_proc_summary = pd.DataFrame(dx_proc_summary).sort_values("diff", ascending=False, key=lambda s: s.abs())

display(dx_proc_summary)
```

STEP 5D — Deteksi outlier (metode IQR) untuk fitur numerik utama

```python
def iqr_outlier_summary(df, column):
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    return {
        "column": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower,
        "upper_bound": upper,
        "n_outliers": len(outliers),
        "pct_outliers": round(len(outliers) / len(df) * 100, 2),
        "fraud_rate_outliers": round(outliers["label"].mean() * 100, 2) if len(outliers) > 0 else np.nan,
        "fraud_rate_non_outliers": round(df.loc[~df.index.isin(outliers.index), "label"].mean() * 100, 2)
    }

outlier_summary = pd.DataFrame([
    iqr_outlier_summary(train, col)
    for col in ["umur", "los", "severitylevel"]
])

display(outlier_summary)
```

STEP 5F — Uji statistik signifikansi (chi-square & Mann-Whitney U)

```python
from scipy.stats import chi2_contingency, mannwhitneyu

chi_results = []

for col in ["typeppk", "jkpst", "cmg", "diagprimer", "kdkc", "dati2"]:
    contingency = pd.crosstab(train[col], train["label"])
    chi2, p_value, dof, expected = chi2_contingency(contingency)

    chi_results.append({
        "column": col,
        "chi2_stat": round(chi2, 2),
        "p_value": p_value,
        "significant (p<0.05)": p_value < 0.05
    })

chi_results = pd.DataFrame(chi_results).sort_values("p_value")

print("=== CHI-SQUARE TEST (kategorikal vs label) ===")
display(chi_results)

mw_results = []

for col in ["umur", "los", "severitylevel", "jnspelsep"]:
    group0 = train.loc[train["label"] == 0, col]
    group1 = train.loc[train["label"] == 1, col]

    stat, p_value = mannwhitneyu(group0, group1, alternative="two-sided")

    mw_results.append({
        "column": col,
        "u_stat": stat,
        "p_value": p_value,
        "significant (p<0.05)": p_value < 0.05
    })

mw_results = pd.DataFrame(mw_results).sort_values("p_value")

print("\n=== MANN-WHITNEY U TEST (numerik vs label) ===")
display(mw_results)
```

STEP 5G — Cross-tabulation antar fitur kategorikal (`typeppk` x `cmg`)

```python
top_typeppk = train["typeppk"].value_counts().head(8).index
top_cmg = train["cmg"].value_counts().head(8).index

subset = train[train["typeppk"].isin(top_typeppk) & train["cmg"].isin(top_cmg)]

print("=== JUMLAH KLAIM: typeppk x cmg ===")
count_table = pd.crosstab(subset["typeppk"], subset["cmg"])
display(count_table)

print("\n=== FRAUD RATE (%): typeppk x cmg ===")
fraud_table = pd.crosstab(
    subset["typeppk"], subset["cmg"],
    values=subset["label"], aggfunc="mean"
) * 100
display(fraud_table.round(1))
```
